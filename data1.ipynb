### Data Format Comparison

| Format | Representation | Storage | Schema Support | Human Readable | Compression | Typical Use | Advantages | Limitations |
|---|---|---|---|---|---|---|---|---|
| CSV | Tabular | Row-based | No | Yes | Low | Simple datasets | Simple and widely supported | Larger files, no schema |
| JSON | Key-value / nested | Text-based | Flexible | Yes | Low | Web and APIs | Supports complex data | Larger file size |
| Parquet | Tabular | Column-based | Yes | No | High | Data analytics | Small size, efficient storage | Not human-readable |
| ORC | Tabular | Column-based | Yes | No | High | Big data | Good compression | Mainly used in big-data systems |
| Avro | Record-based | Binary | Yes | No | Good | Data exchange | Strong schema support | Requires special tools |

### Selected Formats

CSV was selected as the alternative format because it is simple, human-readable, and widely supported.

Parquet was selected as the most suitable format because it supports data types, uses column-based storage, and provides efficient compression. In our experiment, the Parquet file was 22.07 KB, while the CSV file was 41.00 KB.

### Practical Comparison: CSV vs Parquet

| Metric | CSV | Parquet |
|---|---:|---:|
| File Size | 41.00 KB | 22.07 KB |
| Writing Time | 0.007900 s | 0.005642 s |
| Reading Time | 0.011026 s | 0.075419 s |
| Rows | 395 | 395 |
| Columns | 33 | 33 |
| Missing Values | 0 | 0 |
| Data Types | 17 object, 16 int64 | 17 object, 16 int64 |

Both formats preserved all 395 rows, 33 columns, data types, and missing values. Parquet produced a smaller file and was faster to write in this experiment. CSV was faster to read in this particular run.
