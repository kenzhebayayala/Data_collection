TASK 1

A two-dimensional NumPy array with 3 rows and 4 columns was created. I used indexing and slicing to select rows, columns, smaller parts, and individual values. Basic statistics such as mean, median, minimum, maximum, standard deviation, and sum were calculated. Vectorized operations were also performed without using Python loops.  
axis=0 performs calculations down the rows and gives a result for each column, while axis=1 performs calculations across the columns and gives a result for each row.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# Create a two-dimensional NumPy array

```python

arr = np.array([
    [10, 20, 30, 40],
    [15, 25, 35, 45],
    [50, 60, 70, 80]
])

print(arr)
```

# Array structure

```python

print("Shape:", arr.shape)
print("Number of dimensions:", arr.ndim)
print("Size:", arr.size)
print("Data type:", arr.dtype)
```

# Indexing and slicing

```python


print("First row:")
print(arr[0])

print("\nSecond column:")
print(arr[:, 1])

print("\nSmaller part of array:")
print(arr[0:2, 1:3])

print("\nIndividual value:")
print(arr[1, 2])
```

# Basic statistics

```python


print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Minimum:", np.min(arr))
print("Maximum:", np.max(arr))
print("Standard deviation:", np.std(arr))
print("Sum:", np.sum(arr))
```

# Statistics along different axes

```python


print("Mean axis=0:")
print(np.mean(arr, axis=0))

print("\nMean axis=1:")
print(np.mean(arr, axis=1))
```

# Vectorized operations (without Python loops)

```python


print("Add 10:")
print(arr + 10)

print("\nMultiply by 2:")
print(arr * 2)

print("\nSquare:")
print(arr ** 2)

print("\nValues greater than 30:")
print(arr[arr > 30])
```

Task 3

 Indexing
In this task, I used loc and iloc to select rows, columns, ranges of data, and individual values from the DataFrame.

```python

print("Using loc:")
print(df.loc[0:4, ['gender', 'math score', 'reading score']])
```

Using iloc
iloc selects data using integer positions. In this example, I selected the first five rows and specific columns by their positions.

```python

print("Using iloc:")
print(df.iloc[0:5, [0, 5, 6]])
```

Selecting a Range and Individual Value
The following examples show how to select a range of data and a single value using loc and iloc

```python

print("Range using loc:")
print(df.loc[10:14, ['parental level of education', 'math score', 'writing score']])


print("\nIndividual value using loc:")
print(df.loc[10, 'math score'])


print("\nIndividual value using iloc:")
print(df.iloc[10, 5])
```

Difference Between loc and iloc
loc selects data using row and column labels, while iloc selects data using integer positions. Both methods can be used to select rows, columns, ranges, and individual values. In the examples above, both methods selected the same data in different ways.

TASK 4

Question 1: Which students scored more than 80 in math?
This filter selects students whose math score is greater than 80.

```python
high_math = df[df['math score'] > 80]

high_math[['gender', 'math score', 'reading score', 'writing score']].head(10)
```

Question 2: Which students completed the test preparation course?
This filter selects students who completed the test preparation course.

```python
completed_course = df[df['test preparation course'] == 'completed']

completed_course[['gender', 'test preparation course',
                  'math score', 'reading score', 'writing score']].head(10)
```

Question 3: Which students scored more than 70 in both math and reading?
This filter uses multiple conditions to select students whose math and reading scores are both greater than 70.

```python
high_scores = df[
    (df['math score'] > 70) &
    (df['reading score'] > 70)
]

high_scores[['gender', 'math score',
             'reading score', 'writing score']].head(10)
```

Question 4: Which students scored above the average math score?
First, the mean math score is calculated. Then, students whose math score is higher than the mean are selected.

```python
mean_math = df['math score'].mean()

print("Mean math score:", mean_math)

above_average = df[df['math score'] > mean_math]

above_average[['gender', 'math score',
               'reading score', 'writing score']].head(10)
```

Four different filtering methods were used in this task: a numerical condition, a categorical condition, multiple conditions, and a condition based on the mean. The results show how Pandas filtering can be used to select specific groups of students from the dataset.

**Task 5.** Exploratory Data Analysis (EDA)
In this task, I performed an initial exploration of the dataset. I checked data quality, duplicates, categorical distributions, numerical statistics, and relationships between numerical variables.

```python
print("Missing values:")
print(df.isnull().sum())

print("\nNumber of duplicate rows:")
print(df.duplicated().sum())
```

Categorical Distribution
I used value_counts() to examine the distribution of important categorical variables.

```python
print("Gender distribution:")
print(df['gender'].value_counts())

print("\nTest preparation course distribution:")
print(df['test preparation course'].value_counts())

print("\nLunch distribution:")
print(df['lunch'].value_counts())
```

Numerical Distribution
I examined the distributions of math, reading, and writing scores using histograms.

```python
df[['math score', 'reading score', 'writing score']].hist(
    figsize=(10, 6),
    bins=15
)

plt.tight_layout()
plt.show()
```

Correlation Between Scores
I calculated the correlation between math, reading, and writing scores to examine the relationships between these numerical variables.

```python
correlation = df[['math score', 'reading score', 'writing score']].corr()

print("Correlation matrix:")
print(correlation)
```

Five Findings
1. The dataset has no missing values and no duplicate rows.
2. There are slightly more female students (518) than male students (482).
3. Most students (642) did not complete the test preparation course, while 358 students completed it.
4. The average reading score (69.17) is higher than the average math score (66.09) and writing score (68.05).
5. Reading and writing scores have the strongest positive correlation (0.955), which means that students with higher reading scores also tend to have higher writing scores.

Relationship Between Reading and Writing Scores
The scatter plot below shows the relationship between reading and writing scores.

```python
plt.scatter(df['reading score'], df['writing score'], alpha=0.5)

plt.xlabel('Reading Score')
plt.ylabel('Writing Score')
plt.title('Reading Score vs Writing Score')

plt.show()
```
